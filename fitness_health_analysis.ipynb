```python
df = df.drop_duplicates().reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
```

```python
numeric_columns = [
    "Age",
    "Height (cm)",
    "Weight (kg)",
    "Workout Duration (Minutes)",
    "Calories Burned",
    "Steps",
    "Heart Rate (BPM)",
    "Resting Heart Rate (BPM)",
    "Sleep Hours",
    "Water Intake (Liters)",
    "Daily Calories Intake",
    "Workout Frequency"
]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
# ============================================
# DATA CLEANING
# ============================================

# Remove duplicate records
df = df.drop_duplicates().reset_index(drop=True)

# Convert invalid values to NaN

df.loc[
    (df["Age"] < 18) | (df["Age"] > 100),
    "Age"
] = np.nan

df.loc[
    (df["Height (cm)"] < 120) |
    (df["Height (cm)"] > 250),
    "Height (cm)"
] = np.nan

df.loc[
    (df["Weight (kg)"] <= 0) |
    (df["Weight (kg)"] > 300),
    "Weight (kg)"
] = np.nan

df.loc[
    df["Calories Burned"] < 0,
    "Calories Burned"
] = np.nan

df.loc[
    (df["Heart Rate (BPM)"] < 40) |
    (df["Heart Rate (BPM)"] > 220),
    "Heart Rate (BPM)"
] = np.nan

df.loc[
    (df["Resting Heart Rate (BPM)"] < 30) |
    (df["Resting Heart Rate (BPM)"] > 120),
    "Resting Heart Rate (BPM)"
] = np.nan

df.loc[
    (df["Sleep Hours"] < 0) |
    (df["Sleep Hours"] > 24),
    "Sleep Hours"
] = np.nan

df.loc[
    (df["Water Intake (Liters)"] < 0) |
    (df["Water Intake (Liters)"] > 15),
    "Water Intake (Liters)"
] = np.nan


# Fill numerical missing values with median

numeric_columns = [
    "Age",
    "Height (cm)",
    "Weight (kg)",
    "Workout Duration (Minutes)",
    "Calories Burned",
    "Steps",
    "Heart Rate (BPM)",
    "Resting Heart Rate (BPM)",
    "Sleep Hours",
    "Water Intake (Liters)",
    "Daily Calories Intake",
    "Workout Frequency"
]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())


# Convert Activity Date to datetime

df["Activity Date"] = pd.to_datetime(
    df["Activity Date"],
    errors="coerce"
)


# Convert appropriate columns to integers

integer_columns = [
    "Age",
    "Workout Duration (Minutes)",
    "Calories Burned",
    "Steps",
    "Heart Rate (BPM)",
    "Resting Heart Rate (BPM)",
    "Daily Calories Intake",
    "Workout Frequency"
]

for column in integer_columns:
    df[column] = df[column].round().astype(int)

print("Data cleaning completed successfully!")
```

```python
# ============================================
# FINAL DATASET VERIFICATION
# ============================================

print("========== FINAL DATASET VERIFICATION ==========")

print("Shape:", df.shape)

print("\nTotal missing values:",
      df.isnull().sum().sum())

print("\nDuplicate records:",
      df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
# ============================================
# BMI ANALYSIS
# ============================================

# Convert height from centimeters to meters
height_m = df["Height (cm)"] / 100

# Calculate BMI
df["BMI"] = df["Weight (kg)"] / (height_m ** 2)

# Classify BMI
def classify_bmi(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal Weight"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"

df["BMI Category"] = df["BMI"].apply(classify_bmi)

print("BMI calculated successfully!")

display(
    df[
        [
            "User ID",
            "User Name",
            "Height (cm)",
            "Weight (kg)",
            "BMI",
            "BMI Category"
        ]
    ].head(10)
)
```

```python
# ============================================
# BMI STATISTICS
# ============================================

print("========== BMI CATEGORY DISTRIBUTION ==========")

display(
    df["BMI Category"]
    .value_counts()
)

print(
    "\nAverage BMI:",
    round(df["BMI"].mean(), 2)
)

highest_bmi = df.loc[df["BMI"].idxmax()]

print("\n========== HIGHEST BMI ==========")
display(
    highest_bmi[
        ["User ID", "User Name", "BMI", "BMI Category"]
    ].to_frame().T
)

lowest_bmi = df.loc[df["BMI"].idxmin()]

print("========== LOWEST BMI ==========")
display(
    lowest_bmi[
        ["User ID", "User Name", "BMI", "BMI Category"]
    ].to_frame().T
)
```

```python
# ============================================
# BMI GROUP ANALYSIS
# ============================================

print("========== AVERAGE BMI BY GENDER ==========")

display(
    df.groupby("Gender")["BMI"]
    .mean()
    .round(2)
)

print("========== AVERAGE BMI BY CITY ==========")

display(
    df.groupby("City")["BMI"]
    .mean()
    .sort_values(ascending=False)
    .round(2)
)

print("========== AVERAGE CALORIES BY BMI CATEGORY ==========")

display(
    df.groupby("BMI Category", observed=True)["Calories Burned"]
    .mean()
    .round(2)
)

print("========== WORKOUT HABITS BY BMI CATEGORY ==========")

bmi_workout_analysis = df.groupby(
    "BMI Category",
    observed=True
).agg(
    Average_Workout_Duration=("Workout Duration (Minutes)", "mean"),
    Average_Workout_Frequency=("Workout Frequency", "mean"),
    Average_Steps=("Steps", "mean"),
    Average_Calories=("Calories Burned", "mean")
).round(2)

display(bmi_workout_analysis)
```

# 17. Fitness Performance Ranking

```python
# 17. Fitness Performance Ranking

ranking_metrics = [
    "Calories Burned",
    "Steps",
    "Workout Duration (Minutes)",
    "Workout Frequency",
    "Sleep Hours"
]

# Min-Max normalization for each metric
for col in ranking_metrics:
    min_val = df[col].min()
    max_val = df[col].max()

    if max_val != min_val:
        df[f"{col} Score"] = (
            (df[col] - min_val) / (max_val - min_val)
        )
    else:
        df[f"{col} Score"] = 0

# Equal-weight fitness score
df["Fitness Score"] = (
    df["Calories Burned Score"] +
    df["Steps Score"] +
    df["Workout Duration (Minutes) Score"] +
    df["Workout Frequency Score"] +
    df["Sleep Hours Score"]
) / 5 * 100

df["Fitness Score"] = df["Fitness Score"].round(2)

print("Fitness score calculated successfully!")
df[["User ID", "User Name", "Fitness Score"]].head(10)
```

```python
def categorize_fitness_score(score):
    if score < 25:
        return "Low"
    elif score < 50:
        return "Moderate"
    elif score < 75:
        return "Good"
    else:
        return "Excellent"

df["Fitness Score Category"] = df["Fitness Score"].apply(
    categorize_fitness_score
)

print("Fitness score categories:")
display(df["Fitness Score Category"].value_counts())
```

```python
user_ranking = (
    df.groupby(["User ID", "User Name"])
    .agg(
        Average_Fitness_Score=("Fitness Score", "mean"),
        Total_Activities=("Record ID", "count"),
        Average_Calories=("Calories Burned", "mean"),
        Average_Steps=("Steps", "mean"),
        Average_Workout_Duration=("Workout Duration (Minutes)", "mean"),
        Average_Sleep=("Sleep Hours", "mean")
    )
    .reset_index()
)

user_ranking["Average_Fitness_Score"] = (
    user_ranking["Average_Fitness_Score"].round(2)
)

user_ranking = user_ranking.sort_values(
    "Average_Fitness_Score",
    ascending=False
).reset_index(drop=True)

user_ranking["Rank"] = range(1, len(user_ranking) + 1)

print("Top 10 Fitness Users:")
display(user_ranking.head(10))
```

```python
print("Bottom 10 Fitness Users:")
display(user_ranking.tail(10).sort_values("Average_Fitness_Score"))
```

```python
weekend_analysis = (
    df.groupby("Is Weekend")
    .agg(
        Activities=("Record ID", "count"),
        Average_Calories=("Calories Burned", "mean"),
        Average_Steps=("Steps", "mean"),
        Average_Workout_Duration=("Workout Duration (Minutes)", "mean")
    )
    .reset_index()
)

weekend_analysis["Day Type"] = weekend_analysis["Is Weekend"].map({
    True: "Weekend",
    False: "Weekday"
})

print("Weekday vs Weekend Analysis:")
display(
    weekend_analysis[
        [
            "Day Type",
            "Activities",
            "Average_Calories",
            "Average_Steps",
            "Average_Workout_Duration"
        ]
    ]
)
```

```python
day_analysis = (
    df.groupby("Day of Week")
    .agg(
        Activities=("Record ID", "count"),
        Average_Calories=("Calories Burned", "mean"),
        Average_Steps=("Steps", "mean"),
        Average_Workout_Duration=("Workout Duration (Minutes)", "mean")
    )
    .reset_index()
)

day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_analysis["Day of Week"] = pd.Categorical(
    day_analysis["Day of Week"],
    categories=day_order,
    ordered=True
)

day_analysis = day_analysis.sort_values("Day of Week")

print("Activity by Day of Week:")
display(day_analysis)

most_active_day = day_analysis.loc[
    day_analysis["Activities"].idxmax()
]

print(
    f"\nMost Active Day: "
    f"{most_active_day['Day of Week']} "
    f"({most_active_day['Activities']} activities)"
)
```

```python
monthly_activity["Calories Change %"] = (
    monthly_activity["Total_Calories"].pct_change() * 100
).round(2)

monthly_activity["Steps Change %"] = (
    monthly_activity["Average_Steps"].pct_change() * 100
).round(2)

monthly_activity["Workout Duration Change %"] = (
    monthly_activity["Average_Workout_Duration"].pct_change() * 100
).round(2)

print("Monthly Fitness Trends:")
display(monthly_activity)
```
